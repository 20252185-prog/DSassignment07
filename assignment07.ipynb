9.
루트 노드까지 연쇄적으로 영향을 미치려면 항상 부모 노드와 그 형제 노드가 레드여야 한다.
부모 노드(이하 p)와 p의 형제 노드(이하 s)가 모두 레드일 때, 블랙인 p^2와 색깔을 바꾼다. p와 s는 블랙, p^2는 레드가 되었다. 이 때 p^3가 또 레드이기 때문에 성질3가 위배된다.
또한 p^3의 다른 자식인 p^2의 형제 노드 s^2가 레드 노드이고, 그 부모 노드 p^4는 블랙노드이다. p^4는 레드 노드로, p^3과 s^2는 블랙 노드로 바꾼다. 이 때 p^5가 레드 노드이기에 성질3가 위배된다. 
또한 s^3이 레드 노드이고, 그 부모 노드 p^6은 블랙 노드이다. p^6는 레드 노드로, p^5과 s^3는 블랙 노드로 바꾼다. 
이렇게 다음 부모 노드가 레드여서 성질3가 위배되는 동시에 형제 노드 또한 레드인 경우 연쇄적인 수선이 이루어진다.
